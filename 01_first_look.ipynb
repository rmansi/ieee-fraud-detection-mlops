# 01 - First look at the data

Checks the raw CSVs and confirms the Parquet conversion (`src/data/make_parquet.py`) worked.

```python
from pathlib import Path

import pandas as pd
```

```python
print("Notebook is running from:", Path.cwd())

raw = Path("data")
print("Data folder found:", raw.exists())

for f in sorted(raw.glob("*.csv")):
    print(f"{f.name:30s} {f.stat().st_size / 1e6:8.1f} MB")
```

```python
def count_rows(path):
    with open(path) as f:
        return sum(1 for _ in f) - 1  # minus the header line

print("train_transaction rows:", count_rows(raw / "train_transaction.csv"))
print("train_identity rows:   ", count_rows(raw / "train_identity.csv"))
```
